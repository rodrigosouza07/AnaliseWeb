```python
df = df.rename(columns = {'Product ID' : 'Id Produto', 'Product Name':'Nome Produto', 'Listing Price' : 'Lista de preço', 'Sale Price' : 'Preço Promocional', 'Discount' : 'Desconto',
                'Brand' : 'Marca', 'Description' : 'Dezscrição', 'Rating' : 'Avaliação', 'Reviews' : 'Análise', 'Last Visited' : 'Visita mais recente'})

df.head()
```

```python
total_lista = df['Lista de preço'].sum()
total_desconto = df['Preço Promocional'].sum()

desconto = total_lista - total_desconto

print(f"Total sem desconto: R$ {total_lista:,.2f}")
print (f"Total com desconto: R$ {total_desconto:,.2f}")
print(f"Diferença final é: R$ {desconto:,.2f}")
```
