```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

# Load the dataset
df = pd.read_csv("data.csv")
df.head()
```

```python
df.rename(columns={'Student_ID': 'ID_Estudante', 'Academic_Level' : 'Nivel_Academico',  'Age': 'Idade', 'Gender': 'Genero', 'Daily_Usage_Hours': 'Horas_Uso_Diario','Weekend_Extra_Hours': 'Horas_Uso_Fim_Semana', 'Primary_Platform': 'Plataforma_Principal', 'Device_Type': 'Tipo_Dispositivo', 'Sleep_Duration_Hours': 'Duracao_Sono_Horas', 'Sleep_Quality_Score': 'Qualidade_Sono'}, inplace=True)   
df.head()
```

```python
data = df.groupby(['Genero', 'Plataforma_Principal', 'Tipo_Dispositivo'])['Idade'].mean().reset_index()
print(data)
```

```python
# Quantidade de estudantes por gênero e plataforma
uso_por_genero = (
    df.groupby(['Genero', 'Plataforma_Principal'])
      .size()
      .reset_index(name='Quantidade por Genero')
      .sort_values('Quantidade por Genero', ascending=False)
)
display(uso_por_genero)

# Dispositivo mais usado para acessar cada plataforma
uso_dispositivo = (
    df.groupby(['Plataforma_Principal', 'Tipo_Dispositivo'])
      .size()
      .reset_index(name='Quantidade')
      .sort_values('Quantidade', ascending=False)
      .drop_duplicates('Plataforma_Principal')
      .sort_values('Quantidade', ascending=False)
)
display(uso_dispositivo)

# Médias de idade e tempo de uso por plataforma
medias_por_plataforma = (
    df.groupby('Plataforma_Principal')
      .agg(
          Media_Idade=('Idade', 'mean'),
          Media_Horas_Diarias=('Horas_Uso_Diario', 'mean'),
          Media_Horas_Fim_de_Semana=('Horas_Uso_Fim_Semana', 'mean'),
          Genero_Mais_Comum=('Genero', lambda x: x.mode()[0] if not x.mode().empty else np.nan)
      )
      .sort_values('Media_Horas_Diarias', ascending=False)
      .round(2)
      .reset_index()
)
display(medias_por_plataforma)
```

```python
# Dispersão: horas médias de uso diário versus fim de semana
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=medias_por_plataforma,
    x='Media_Horas_Diarias',
    y='Media_Horas_Fim_de_Semana',
    s=100
)

for _, linha in medias_por_plataforma.iterrows():
    plt.annotate(
        linha['Plataforma_Principal'],
        (linha['Media_Horas_Diarias'], linha['Media_Horas_Fim_de_Semana']),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.title('Uso médio por plataforma: dias úteis e fim de semana')
plt.xlabel('Horas médias de uso diário')
plt.ylabel('Horas médias de uso no fim de semana')
plt.tight_layout()
plt.show()


# Barras agrupadas: médias de uso por plataforma
medias_uso = medias_por_plataforma.set_index('Plataforma_Principal')[
    ['Media_Horas_Diarias', 'Media_Horas_Fim_de_Semana']

]

#comparando o uso diário e o uso no fim de semana por plataforma
ax = medias_uso.plot(kind='bar', figsize=(10, 6))
ax.set_title('Horas médias de uso por plataforma')
ax.set_xlabel('Plataforma')
ax.set_ylabel('Horas médias')
ax.legend(['Uso diário', 'Uso no fim de semana'])
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

#comparando o uso diário e o uso no fim de semana por plataforma (empilhado)
ax = medias_uso.plot(kind='bar', figsize=(10, 6), stacked=True)
ax.set_title('Horas médias de uso por plataforma (empilhado)')
ax.set_xlabel('Plataforma')
ax.set_ylabel('Horas médias')
ax.legend(['Uso diário', 'Uso no fim de semana'])
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
